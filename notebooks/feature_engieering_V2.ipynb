```python
Delete_cols = ['id', 'stock_no', 'seller_name', 'street']

df.drop(columns=Delete_cols, inplace=True)
```

```python
df.drop_duplicates(inplace = True)
```

```python
(df['price'] > 150000).value_counts()
```

```python
df.drop_duplicates(subset="vin", keep="first", inplace=True)
```

```python
for threshold in [150_000, 250_000, 500_000]:
    print(
        threshold,
        (df["price"] > threshold).sum()
    )
```

```python
Expensive_cars = df.loc[
    df["price"] > 250_000,
    ["vin","price", "year", "make", "model", "trim", "miles"]
]
```

```python
(Expensive_cars['make']).value_counts()
```

```python
(Expensive_cars['vin']).nunique()
```
